# Retrieval-Augmented Generation Demo

This notebook compares a plain prompt with context retrieved via `HierarchicalService`. The dataset lives in `examples/data/sample_docs.json`.

```
import json
from pathlib import Path
from deepthought.search import OfflineSearch
from deepthought.services import HierarchicalService

class DummyNATS:
    pass

class DummyJS:
    pass

path = Path('examples/data/sample_docs.json')
examples = json.loads(path.read_text())
index = OfflineSearch.create_index('rag_demo.db', [(d['title'], d['content']) for d in examples])
service = HierarchicalService(DummyNATS(), DummyJS(), None, search=index)
```

```
question = 'What is SQLite?'
print('Plain question:', question)
```

```
ctx = service.retrieve_context(question)
print('Retrieved context:', ctx)
```

Use the context to build a prompt for your LLM:

```
prompt = f'Question: {question}
Context: {ctx[0]}
Answer:'
print(prompt)
```
